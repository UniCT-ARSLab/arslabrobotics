```python
#
# TO BE USED WITH godot/cart2D_no_physics
#

import arslabrobotics as ar

dds = ar.dds.DDS()
dds.start()

#dds.subscribe(['tick'])

cart2d = ar.system.TwoWheelsCart2D(1.0, 0.15, 0.9, 0.9, 0.3)


t = ar.utils.Time()
t.start()
while t.get() < 3:

    #dds.wait('tick')
    t.sleep(0.02)
    delta_t = t.elapsed()
    
    cart2d.evaluate(delta_t, 1, 1)
    pose = cart2d.get_pose()

    dds.publish('X', pose[0], ar.dds.DDS_TYPE_FLOAT)
    dds.publish('Y', pose[1], ar.dds.DDS_TYPE_FLOAT)
    dds.publish('Theta', pose[2], ar.dds.DDS_TYPE_FLOAT)

dds.stop()
```
