```python

import arslabrobotics as ar
import numpy as np

lin_sys = ar.system.LinearDynamicSystem( [ [-3, -2],
                                 [1, 0,] ],
                               [[1], [0]],
                               [[0, 1]])

#print()
#print(lin_sys.B @ lin_sys.C)

kp = 0
while kp > -10:
    new_a = lin_sys.A - kp * (lin_sys.B @ lin_sys.C)
    #print(new_a, lin_sys.B @ lin_sys.C)
    print(kp, np.linalg.eig(new_a).eigenvalues)
    kp -= 0.01
```
