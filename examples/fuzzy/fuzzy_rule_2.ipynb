```python
#
# fuzzy_rule_2.py
#

import arslabrobotics as ar

hot = ar.fuzzy.Up("Hot", 25, 50)
warm = ar.fuzzy.Triangle("Warm", 15, 30)
cold = ar.fuzzy.Down("Cold", 10, 20)
ar.fuzzy.plot_fuzzysets([cold, warm, hot])

fast = ar.fuzzy.Up("Fast", 70, 100)
mid = ar.fuzzy.Triangle("Mid", 20, 80)
slow = ar.fuzzy.Down("Slow", 0, 30)
ar.fuzzy.plot_fuzzysets([fast,mid,slow])
```

```python
t = 17  # put temperature here
temperature = ar.fuzzy.Variable(t)
print("Hot fuzzy value:", temperature(hot).get())

fan = ar.fuzzy.Variable()

r1 = (temperature(hot)) >> fan(fast)
r2 = (temperature(warm)) >> fan(mid)
r3 = (temperature(cold)) >> fan(slow)

e = ar.fuzzy.FuzzyEngine()
e.use_cog()
e.plot = True
e.execute(r1,r2,r3)
print("fan speed:", e.get(fan))
```
