```python
#
#
#

import arslabrobotics as ar

hot = ar.fuzzy.Up("Hot", 30, 40)
ar.fuzzy.plot_fuzzysets([hot])

fast = ar.fuzzy.Up("Fast", 80, 100)
ar.fuzzy.plot_fuzzysets([fast])
```

```python
t = 32  # put temperature here
temperature = ar.fuzzy.Variable(t)
print("Hot fuzzy value:", temperature(hot).get())

fan = ar.fuzzy.Variable()

r1 = (temperature(hot)) >> fan(fast)

e = ar.fuzzy.FuzzyEngine()
e.plot = True
e.execute(r1)
print("fan speed:", e.get(fan))
```
