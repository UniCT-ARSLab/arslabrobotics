```python
import arslabrobotics as ar

dps = ar.data.DataPlotter()
dps.set_x("time (seconds)")
dps.add_y("speed", "speed")

dpa = ar.data.DataPlotter()
dpa.set_x("time (seconds)")
dpa.add_y("angle", "Angle")

arm = ar.system.Arm(1.0, 0.5, 0.9)
# mass = 1 Kg
# length = 0.5 meters
# friction = 0.9

torque = 2.5 # Nm

t = ar.utils.Time(use_fake_time=True)
t.start()
while t.get() < 20:

    t.sleep(0.01)
    delta_t = t.elapsed()
    (p, v) = arm.evaluate(delta_t, torque)

    dps.append_x(t.get())
    dps.append_y("speed", v)
    
    dpa.append_x(t.get())
    dpa.append_y("angle", p)

dpa.plot()
dps.plot()

```
