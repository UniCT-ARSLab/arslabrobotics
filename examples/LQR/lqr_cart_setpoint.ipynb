```python
#
#
#

import arslabrobotics as ar

import numpy as np
import scipy

# continuous time matrices
A = np.array( [ [-0.9, 0],
                [1, 0] ])
B = np.array([ [1], [0] ])

delta_t = 0.1  # 100ms

# discrete time matrices
A = np.eye(2) + A * delta_t
B = B * delta_t

print("State Matrix:\n", A)
print("Input Matrix:\n", B)
```

```python

Q = np.eye( 2, dtype=np.float32) * 1
Q[0] = 0
R = np.eye( 1, dtype=np.float32) * 0.1

P = scipy.linalg.solve_discrete_are(A, B, Q, R)
K = np.linalg.inv(R + B.T @ P @ B) @ B.T @ P @ A

print("Optimal Gain:", K)
```

```python

#initial state
x = np.array([ 0., 0.])

set_point = np.array([0., 10.])

#x_tr = x - set_point

dps = ar.data.DataPlotter()
dps.set_x("steps")
dps.add_y("speed", "speed")

dpp = ar.data.DataPlotter()
dpp.set_x("steps")
dpp.add_y("pos", "position")

for t in range(0,200):
    #u = - K @ x_tr
    #x_tr = A @ x_tr + B @ u
    #x = set_point + x_tr
    u = -K @ (x - set_point)
    x = A @ x + B @ u
    dps.append_x(t)
    dps.append_y("speed", x[0])
    dpp.append_x(t)
    dpp.append_y("pos", x[1])

dps.plot()
dpp.plot()
```
