```python
#
#
#
import numpy as np
import matplotlib.pyplot as plt
import arslabrobotics as ar

delta_t = 0.5e-4  # Sampling interval 100us f = 20KHz
freq = 2000

signal = []
w = 2 * np.pi * freq 

# generate a signal using Fourier series
tm = ar.utils.Time(use_fake_time = True)
tm.start()
while tm.get() < 1:
    tm.sleep(delta_t)
    tm.elapsed()
    val = 10 * np.sin(w*tm.get())
    signal.append(val)

n_samples = 100 #len(signal)
plt.plot(np.arange(n_samples) * delta_t, signal[:n_samples])
plt.title('Time Domain')
plt.xlabel('Time')
plt.ylabel('Magnitude')
plt.show()
```

```python
# Compute FFT
fft_values = np.fft.fft(signal)
N = len(signal)

# Compute frequency bins
freq = np.fft.fftfreq(N, delta_t)

# Take the magnitude (absolute value) of the FFT
magnitude = np.abs(fft_values)

n_samples = N//2
plt.plot(freq[:n_samples], magnitude[:n_samples])
plt.title('Frequency Domain (Magnitude Spectrum)')
plt.xlabel('Frequency [Hz]')
plt.ylabel('Magnitude')
plt.show()
```
