```python
#
import numpy as np
import scipy.optimize as opt

Q = np.array([ [2, -3],
               [-3, 2] ])
c = np.array([ [-1], [-2] ])

def func(x):
    x = np.array([x]).T # make "X" a column vector
    y = 0.5 * x.T @ Q @ x + c.T @ x
    return y[0]

# constraint (inequalities are treated as >= 0)
constr = [ {'type':'ineq', 'fun' : lambda x : 8 - (x[0] + x[1]) } ]

# bounds
bnds = [ (0, None), (0, None) ]

# initial guess
x0 = np.zeros(2)

result = opt.minimize(func, x0, constraints=constr, bounds = bnds)
print(result.message, result.x, result.fun)
```
