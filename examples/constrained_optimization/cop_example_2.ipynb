```python
#

import scipy.optimize as opt

# find maximum area of a rectangle with 8m of perimeter

# x[0] = h
# x[1] = w
def func(x):
    return -(x[0] * x[1])  # minus is because we want to maximize

# constraint
constr = [ {'type':'eq', 'fun' : lambda x : 2*(x[0] + x[1]) - 8 } ]

# initial guess
x0 = [ 0.0,  0.0]

result = opt.minimize(func, x0, constraints=constr)
print(result.message, result.x, result.fun)
```
